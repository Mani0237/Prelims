```python
import random

# Function to simulate a single toss of both coins
def toss_coins():
    return random.choice(['Heads', 'Tails'])

# Function to conduct the experiment multiple times
def conduct_experiment(num_trials):
    results = {'Heads': 0, 'Tails': 0}
    for _ in range(num_trials):
        outcome_A = toss_coins()
        outcome_B = toss_coins()
        results[(outcome_A, outcome_B)] = results.get((outcome_A, outcome_B), 0) + 1
    return results

# Main function
def main():
    num_trials = 100000
    results = conduct_experiment(num_trials)
    print("Results of {} trials:".format(num_trials))
    print("Number of times both coins landed on Heads:", results.get(('Heads', 'Heads'), 0))
    print("Number of times both coins landed on Tails:", results.get(('Tails', 'Tails'), 0))

if __name__ == "__main__":
    main()
```

```python
# Function to simulate a single toss of a given coin
def toss_coin(coin):
    if coin == 'Heads':
        return 'Heads'
    elif coin == 'Tails':
        return 'Tails'

# Function to conduct the experiment multiple times
def conduct_experiment(num_trials):
    results = {'Heads': 0, 'Tails': 0}
    for _ in range(num_trials):
        # Randomly select a coin (Coin A or Coin B)
        coin = random.choice(['Heads', 'Tails'])
        outcome = toss_coin(coin)
        results[outcome] += 1
    return results

# Main function
def main():
    num_trials = 10000
    results = conduct_experiment(num_trials)
    print("Results of {} trials:".format(num_trials))
    print("Number of times Heads was observed:", results['Heads'])
    print("Number of times Tails was observed:", results['Tails'])

if __name__ == "__main__":
    main()
```
