```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from scipy.integrate import solve_ivp
```

```python
# defining the function
def f(y):
    return 5 * np.sinh(5*y) 
```

### Part (a)

Converting the second order ODE to a system of two first order ODEs
$$
\dfrac{dy}{dt} = z
$$
$$
\dfrac{dz}{dt} = \lambda sinh(\lambda y)
$$

### Part (b)

```python
# Define the system of first-order ODEs
def ode(t, y, lam):
    dydt = y[1]
    dzdt = lam * np.sinh(lam * y[0])
    return [dydt, dzdt]

# Define parameters
lam = 5.0
s = 0.01

# Define time span
t_span = [0, 1]

# Initial conditions
y0 = 0
z0 = s
u0 = [y0, z0]

# Solve the IVP
sol = solve_ivp(lambda t, y: ode(t, y, lam), t_span, u0, method='RK45', t_eval=np.linspace(0, 1, 100))

# Extracting time and y values
t_values = sol.t
y_values = sol.y[0]

# Plot y vs t
plt.plot(t_values, y_values, label='y vs t')
plt.xlabel('t')
plt.ylabel('y')
plt.title('Solution of y vs t')
plt.grid(True)
plt.legend()
plt.show()

# Report y(1)
y_at_1 = sol.y[0][-1]
print("Solution for y(1) with s = 0.01:", y_at_1)
```
