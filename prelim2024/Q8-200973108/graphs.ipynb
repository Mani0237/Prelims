### OpenMP

```python
import matplotlib.pyplot as plt

# Integration results
threads = [1, 2, 4, 6, 8, 10, 16, 28]
time_microseconds = [20035, 10570, 5327, 3635, 2723, 2225, 1578, 1056]

# Calculate scalability
speedup = [time_microseconds[0] / t for t in time_microseconds]

# Create subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot integration time
ax1.plot(threads, time_microseconds, marker='o', color='maroon', linestyle='-')
ax1.set_xlabel('Threads')
ax1.set_ylabel('Time (microseconds)')
ax1.set_title('Integration Time vs. Number of Threads')
ax1.minorticks_on()
ax1.grid(True, which='both', linestyle=':', linewidth='0.5', color='gray', alpha=0.7)
ax1.set_xticks(threads)

# Plot scalability
ax2.plot(threads, speedup, marker='o', color='green', linestyle='-')
ax2.set_xlabel('Threads')
ax2.set_ylabel('Speedup')
ax2.set_title('Speedup vs. Number of Threads')
ax2.minorticks_on()
ax2.grid(True, which='both', linestyle=':', linewidth='0.5', color='gray', alpha=0.7)
ax2.set_xticks(threads)

# Adjust layout and display
plt.tight_layout()
plt.show()
```
