```python
import matplotlib.pyplot as plt
```

```python
# Data for threads and scalability
threads = [1, 2, 4, 8, 16, 28]

scalability_static = [1.00, 1.82, 3.53, 7.03, 8.04, 13.23]
scalability_dynamic = [1.00, 1.81, 3.55, 7.12, 12.08, 13.44]
scalability_guided = [1.00, 1.83, 3.62, 7.19, 11.83, 13.51]

# Plotting the data
plt.figure(figsize=(8, 5))
plt.plot(threads, scalability_static, marker='o', label='Static', color='blue')
plt.plot(threads, scalability_dynamic, marker='s', label='Dynamic', color='green')
plt.plot(threads, scalability_guided, marker='^', label='Guided', color='maroon')

# Adding titles and labels
plt.title('Scalability vs Threads')
plt.xlabel('Number of Threads')
plt.ylabel('Scalability')
plt.legend()
plt.minorticks_on()
plt.grid(True, which='both', linestyle=':', linewidth='0.5', color='gray', alpha=0.7)

# Display the plot
plt.show()
```
