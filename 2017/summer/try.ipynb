```python
import numpy as np 
 
def f(x):
    return x + 5*np.power(x,2) - 0.5*np.power(x,3)

def trapezoid(a, b, n):
    h = (b - a)/n 
    sum = f(a) + f(b)
    
    for i in range(1,n):
        x = a + i*h 
        sum = sum + f(x)

    return h*sum      

a = 0
b = 10
n = 10000
result = trapezoid(a,b,n)
print("result = ", result)
```
