```python
# importing libraries
import numpy as np
import pprint as pp
import matplotlib.pyplot as plt
```

### Part (b)

```python
def generate_A(N):
    '''------------------------------------------
    @brief - This function generates a tridiagonal matrix A.
    
    @param N - size of tridiagonal matrix
    @return A tridiagonal matrix
    ---------------------------------------------'''
    
    A = np.zeros((N,N))
    
    for i in range(N):
        # main diagonal
        A[i][i] = 2
        
        # superdiagonal
        if i < N - 1:
            A[i][i+1] = -1
        
        # subdiagonal
        if i > 0:
            A[i][i-1] = -1
    
    return A
```

```python
N = 8
A = generate_A(N)

# printing matrix
print("Matrix A:")
pp.pprint(A)

B = np.linalg.inv(A)
B = B * ((N+1))
print("Matrix B:")
pp.pprint(B)
```

### Part (c)

```python
# Function to extract the Bij(k) matrix
def extract_block(B, i, j, k):
    """
    Extracts the Bij(k) matrix from the partitioned matrix B.
    
    Args:
    - B: Partitioned matrix B
    - i, j: Indices of the block within the partitioned matrix B
    - k: Level of the block within the partitioning
    
    Returns:
    - Block matrix Bij(k)
    """
    # Calculate the size of the blocks at level k
    block_size = B.shape[0] // (2 ** k)
    
    # Calculate the starting and ending row indices of the block
    start_row = (i * block_size)
    end_row = start_row + block_size
    
    # Calculate the starting and ending column indices of the block
    start_col = (j * block_size)
    end_col = start_col + block_size
    
    # Extract the block from the partitioned matrix B
    block = B[start_row:end_row, start_col:end_col]
    
    return block


# Define the indices for the blocks
blocks = [
    ("11", 2), ("12", 2),
    ("21", 2), ("22", 2),
    ("33", 2), ("34", 2),
    ("43", 2), ("44", 2),
    ("12", 1), ("21", 1)
]

# Extract and print each block matrix
for block_index, level in blocks:
    block = extract_block(B, int(block_index[0]) - 1, int(block_index[1]) - 1, level)
    print(f"B_{block_index}^({level}) matrix:")
    pp.pprint(block)
    print("\n")
```

### Part (d)

```python
# SVD of B_12_1
B_12_1 = extract_block(B, 0, 1, 1)
print("B_12^1: ")
pp.pprint(B_12_1)

U1, S1, V1 = np.linalg.svd(B_12_1)

print("\n")
print("U: ", U1)
print("\n")
print("V: ", V1)
print("\n")
print("S: ", S1)
print("\n")

# SVD of B_12_2
B_12_2 = extract_block(B, 0, 1, 2)
print("B_12^2: ")
pp.pprint(B_12_2)

U2, S2, V2 = np.linalg.svd(B_12_2)

print("\n")
print("U: ", U2)
print("\n")
print("V: ", V2)
print("\n")
print("S: ", S2)
```
