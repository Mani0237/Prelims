```python
# importing libraries
import numpy as np
```

```python
def Power_method_max(A, x0, iter_max, tol):
    """
    Find the largest eigenvalue and eigenvector of a matrix A using the Power Method.

    Parameters:
    - A (numpy.ndarray): The square matrix for which to find the largest eigenvalue and eigenvector.
    - x0 (numpy.ndarray): The initial guess for the eigenvector.
    - iter_max (int): The maximum number of iterations.
    - tol (float): The convergence tolerance.

    Returns:
    - eigenvector (numpy.ndarray): The estimated eigenvector (normalized).
    - eigenvalue (float): The estimated largest eigenvalue.
    - iterations (int): The number of iterations performed.
    """
    prev_eigenvalue = 10

    # Normalize the initial guess
    #x0 = x0 / np.linalg.norm(x0, ord=np.inf)

    for i in range(iter_max):
        x0 = np.dot(A, x0)

        # Scaling the eigenvector
        delta = np.linalg.norm(x0, ord=np.inf)
        x0 = x0 / delta

        # Calculate the corresponding eigenvalue
        eigenvalue = (x0.transpose() @ A @ x0) / (x0.transpose() @ x0)
        
        # Check for convergence
        delta_lambda = np.abs(eigenvalue - prev_eigenvalue)
        print(f"Iteration:{i + 1}  Eigenvalue:{eigenvalue}  Error:{delta_lambda}") 
        if delta_lambda < tol:
            break

        prev_eigenvalue = eigenvalue

    return x0, eigenvalue, i + 1
```

```python
# Define the matrices A1, A2, and A3
A1 = np.array([[1, 1, 1], [1, 2, 3], [1, 3, 6]])

# Initial guess for eigenvector
initial_guess = np.array([1, 1, 1])

# Apply Power Method to find dominant eigenvalues and eigenvectors
eigenvector1, eigenvalue1, iterations1 = Power_method_max(A1, initial_guess, 100, 1e-6)

# Print results
print("Matrix A1:")
print("Dominant Eigenvalue:", eigenvalue1)
print("Dominant Eigenvector:", eigenvector1)
print("Number of Iterations:", iterations1)
print()
```
