```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

### Part 1
We need to plot the function 
$$
\pi(x) = \dfrac{1}{2}\mathcal{N}(1,2) + \dfrac{1}{2}\mathcal{N}(2,2)
$$

```python
sigma = 2
xmin, xmax = np.array([-3*sigma, 4*sigma])

# generating the data of the distribution
x = np.linspace(xmin,xmax,100)
def pi(x): 
    return 0.5 * norm.pdf(x,1,sigma) + 0.5 * norm.pdf(x,2,sigma)

plt.plot(x, pi(x), color='maroon')
plt.xlabel('x')
plt.ylabel(r'$\pi(x)$')
plt.grid(True,linestyle='--',alpha=0.5)
```

### Part 3

```python
# defining input parameters
pi_y = np.array([0.5,0.5])
mu_x = np.array([1, 2])
sigma_values = [1,4]
x_values = np.linspace(-15, 15, 1000)

# computing posterior probability distribution 
for sigma in sigma_values:
    posterior = np.zeros_like(x_values)
    for y, pi_y_value in enumerate(pi_y):
        likelihood = norm.pdf(x_values, mu_x[y],sigma)
        posterior += pi_y_value * likelihood
        
    # normalizing the posterior
    posterior = posterior/np.sum(posterior)
    
    # plotting the posterior probability distribution
    plt.plot(x_values, posterior, label=f'$\sigma$ = {sigma}')

plt.xlabel('x')
plt.ylabel('Posterior Probability $\pi(x)$')
plt.legend()
plt.grid(True,linestyle='--', alpha = 0.5)
plt.show()
    
```
