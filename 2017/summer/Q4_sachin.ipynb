```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

### Part 1
We need to plot the function 
$$
\pi(x) = \dfrac{1}{2}\mathcal{N}(1,2) + \mathcal{N}(2,2)
$$

```python
sigma = 2
xmin, xmax = np.array([-3*sigma, 4*sigma])

# generating the data of the distribution
x = np.linspace(xmin,xmax,100)
def pi(x): 
    return 0.5 * norm.pdf(x,1,sigma) + 0.5 * norm.pdf(x,2,sigma)

plt.plot(x, pi(x), color='maroon')
plt.xlabel('x')
plt.ylabel(r'$\pi(x)$')
plt.grid(True,linestyle='--',alpha=0.5)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
pi_y = np.array([0.5, 0.5])  # probability distribution of y
mu_x = np.array([1, 2])       # mean of x given y
sigma_values = [1, 4]         # standard deviation values for x given y
x_values = np.linspace(-10, 15, 1000)  # range of x values for plotting

# Compute posterior probability distribution for each sigma value
for sigma in sigma_values:
    posterior = np.zeros_like(x_values)
    for y, pi_y_value in enumerate(pi_y):
        likelihood = norm.pdf(x_values, mu_x[y], sigma)
        posterior += pi_y_value * likelihood
    
    # Normalize the posterior
    posterior /= np.sum(posterior)

    # Plot the posterior probability distribution
    plt.plot(x_values, posterior, label=f'σ = {sigma}')

plt.xlabel('x')
plt.ylabel('Posterior Probability π(y|x)')
plt.title('Posterior Probability Distribution for Different σ')
plt.legend()
plt.grid(True)
plt.show()
```
