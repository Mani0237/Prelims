```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# defining function for Power Method
def Power_method_Rayleigh(A, x0, tol, max_iter):
    x = x0 
    eigenvalue = 0
    eigenvector = np.zeros_like(x)
    print("Iteration \t Eigenvalue \t \t EigenVector \t \t \t Convergence Rate")
    print("------------------------------------------------------------------------------")
    prev_error = 0
    for i in range(max_iter):
        x_prev = x 
        
        # Rayleigh Quotient iteration
        x = A @ x 
        # normalization of eigenvector  
        x = x / np.max(np.abs(x))
        
        eigenvalue = x.T @ A @ x 
        eigenvalue = eigenvalue/np.dot(x.T, x)
        
        # computing error 
        error = np.linalg.norm(x - x_prev, np.inf)
        
        # computing convergence
        if i > 0:
            conv_rate = error / prev_error
        else:
            conv_rate = np.nan 

        # Display iteration information
        print(f"{i+1}\t\t{eigenvalue:.8f}\t\t{np.array2string(x, precision=4, separator=',')}\t \t{conv_rate:.8f}")            
        
        # Check for convergence
        if error < tol:
            print(f"\nConverged in {i+1} iterations")
            break
        
        # Update previous error
        prev_error = error
    
    return eigenvalue, x
        
```

### Part 1


```python
# Define the matrices A1, A2, A3
A1 = np.array([[1, 1, 1], [1, 2, 3], [1, 3, 6]])
A2 = np.array([[3, 3, 0], [3, 5, 0], [0, 0, 7]])
A3 = np.array([[3, 1, 0], [0, 3, 0], [0, 0, 2]])

# calculating theoretical eigenvalues
e1, ev1 = np.linalg.eig(A1)
print("Eigenvalues of A1: \n", e1)
e2, ev2 = np.linalg.eig(A2)
print("\nEigenvalues of A2: \n", e2)
e3, ev3 = np.linalg.eig(A3)
print("\nEigenvalues of A3: \n", e3)
```

```python
# Define initial guess
x0 = np.array([1, 1, 1])
tol = 10e-6
max_iter = 100 

# Apply Power Method to A1
print("\nMatrix A1:")
eigenvalue_A1, eigenvector_A1 = Power_method_Rayleigh(A1, x0, tol, max_iter)
print("\nDominant Eigenvalue:", eigenvalue_A1)
print("Dominant Eigenvector:", eigenvector_A1)

# Apply Power Method to A2
print("\nMatrix A2:")
eigenvalue_A2, eigenvector_A2 = Power_method_Rayleigh(A2, x0, tol, max_iter)
print("\nDominant Eigenvalue:", eigenvalue_A2)
print("Dominant Eigenvector:", eigenvector_A2)

# Apply Power Method to A3
print("\nMatrix A3:")
eigenvalue_A3, eigenvector_A3 = Power_method_Rayleigh(A3, x0, tol, max_iter)
print("\nDominant Eigenvalue:", eigenvalue_A3)
print("Dominant Eigenvector:", eigenvector_A3)
```
