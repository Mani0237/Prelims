```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Function to perform Power Method
def power_method(A, x0, tol=1e-6, max_iter=100):
    x = x0
    eigenvalue = 0
    eigenvec = np.zeros_like(x)
    
    print("Iteration\tEigenvalue\t\tEigenVector\t \tConvergence Rate")
    print("--------------------------------------------------------------")
    prev_diff = 0
    for i in range(max_iter):
        x_prev = x
        
        # Power iteration
        x = A @ x
        eigenvalue = np.dot(x, x_prev)
        eigenvalue /= np.dot(x_prev, x_prev)
        
        # Normalize eigenvector
        x = x / np.max(np.abs(x))
        
        # Compute error
        error = np.linalg.norm(x - x_prev, np.inf)
        
        # Compute convergence rate
        if i > 0:
            convergence_rate = error / prev_diff
        else:
            convergence_rate = np.nan
        
        # Display iteration information
        print(f"{i+1}\t\t{eigenvalue:.8f}\t\t{x}\t \t{convergence_rate}")
        
        # Check for convergence
        if error < tol:
            print(f"\nConverged in {i+1} iterations")
            break
        
        # Update previous difference
        prev_diff = error
    
    return eigenvalue, x


```

### Part 1

```python
# Define the matrices A1, A2, A3
A1 = np.array([[1, 1, 1], [1, 2, 3], [1, 3, 6]])
A2 = np.array([[3, 3, 0], [3, 5, 0], [0, 0, 7]])
A3 = np.array([[3, 1, 0], [0, 3, 0], [0, 0, 2]])

# Define initial guess
x0 = np.array([1, 1, 1])

# Apply Power Method to A1
print("\nMatrix A1:")
eigenvalue_A1, eigenvector_A1 = power_method(A1, x0)
print("\nDominant Eigenvalue:", eigenvalue_A1)
print("Dominant Eigenvector:", eigenvector_A1)

# Apply Power Method to A2
print("\nMatrix A2:")
eigenvalue_A2, eigenvector_A2 = power_method(A2, x0)
print("\nDominant Eigenvalue:", eigenvalue_A2)
print("Dominant Eigenvector:", eigenvector_A2)

# Apply Power Method to A3
print("\nMatrix A3:")
eigenvalue_A3, eigenvector_A3 = power_method(A3, x0)
print("\nDominant Eigenvalue:", eigenvalue_A3)
print("Dominant Eigenvector:", eigenvector_A3)
```
