```python
import numpy as np
import matplotlib.pyplot as plt

N = np.array([10, 100, 1000, 10000, 100000, 1000000])
error = np.array([1.30473, 0.203805, 0.0273299, 0.00493167 , 0.00139746 , 0.000232317 ])
sqrt_N = 1 / np.sqrt(N)

plt.loglog(N, error, label='Error', marker='*')
plt.loglog(N, sqrt_N, label='1/sqrt(N)')
plt.xlabel('Number of Samples (N)')
plt.ylabel('Error')
plt.title('Error vs. Number of Samples')
plt.legend()
plt.show()
```
