# Ordinary differential equation

### Part 1: Case N = 6

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# function to generate central difference matrix 
# this matrix is (N-1)*(N-1) since the function is zero at boundries

def second_derivative(a, b, N):
    h = (b-a)/N 
    matrix = -(1.0/h**2)*( np.diag(np.ones(N-2), -1) + np.diag(np.ones(N-2), 1) - 2*np.diag(np.ones(N-1),0))

    return matrix  

A = second_derivative(0, 10, 6)    
print(A)
```

### Part 2: Case N = 200

```python
a = 0
b = 10 
N = 200

# discretizing domain into N + 1 points
x = np.linspace(a, b, N+1)
A = second_derivative(a, b, N)

# finding smallest eigenvalues
eigenvalues, eigenvectors = np.linalg.eigh(A)
Emin = min(eigenvalues)

# get index of smallest eigenvalue
for i in range(len(eigenvalues)):
    if eigenvalues[i] == Emin:
        Emin_index = i 

# get corresponding eigenvector
E_vec = eigenvectors[:,Emin_index]

# plot the solution
plt.plot(x[1:-1], E_vec, 'maroon')
plt.grid(True)
```
