```python
import numpy as np
import matplotlib.pyplot as plt

n = [100, 1000, 10000, 100000, 1000000]
e = [0.74879, 0.14079, 0.0080098, 0.0061502, 0.0060338]

plt.loglog(n, e)
plt.plot(n, 1/np.sqrt(n))

plt.xlabel('Number of Samples (n)')
plt.ylabel('Error (e)')
plt.legend(['Error', '1/√N'])

plt.title('Error vs Number of Samples')
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = [100, 1000, 10000, 100000, 1000000]
e = [0.909164, 0.055576, 0.0368191, 0.00734756, 0.00635982]

plt.loglog(n, e, 'k*')
plt.plot(n, 1/np.sqrt(n))

plt.xlabel('Number of Samples (n)')
plt.ylabel('Error (e)')
plt.legend(['Error', '1/√N'])

plt.title('Error vs Number of Samples')
plt.grid(True)

plt.show()
```
