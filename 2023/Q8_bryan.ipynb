```python
# importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pprint as pp
from scipy.linalg import qr
```

```python
# defining matrix A1
A1 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [7, 10, 10, 3, 2, 7]])

# defining matrix A2
A2 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [-2, 10**(-5), -8, 7, -3, 6]])

# defining matrix b
b = np.array([[-1,0,5]])
```

### Part (b)

```python
def pseudo_inverse(matrix):
    A_AT = matrix @ matrix.T
    return matrix.T @ np.linalg.inv(A_AT)

def calculate_x(A, b):
    A_dagger = pseudo_inverse(A)
    x = A_dagger @ b.T
    residual = np.linalg.norm(A@x-b.T)
    condition_number = np.linalg.cond(A)
    
    return x, residual, condition_number

# results for A1
x1, residual1, cond1 = calculate_x(A1, b)
print("Results for matrix A1: ")
print("--------------------------------")
print("x: \n", x1)
print("\n Residual: \n", residual1)
print("\n Condition number: \n", cond1)

# results for A2
x2, residual2, cond2 = calculate_x(A2, b)
print("\nResults for matrix A2: ")
print("--------------------------------")
print("x: \n", x2)
print("\n Residual: \n", residual2)
print("\n Condition number: \n", cond2)
```

### Part (c)

```python
def economy_inverse(matrix):
    U, s, V = np.linalg.svd(matrix, full_matrices=False) # economy mode of SVD
    S_inv = np.diag(1.0/s)
    return V.T @ S_inv @ U.T 

def calculate_x_economy(matrix, b):
    M = economy_inverse(matrix)
    x = M @ b.T 
    residual = np.linalg.norm(matrix @ x-b.T)
    
    return x, residual

# results for A1
x1, residual1 = calculate_x_economy(A1, b)
print("Results for matrix A1: ")
print("--------------------------------")
print("x: \n", x1)
print("\n Residual: \n", residual1)

# results for A2
x2, residual2 = calculate_x_economy(A2, b)
print("\nResults for matrix A2: ")
print("--------------------------------")
print("x: \n", x2)
print("\n Residual: \n", residual2)
```

### Part (d)

```python
def solve_qr_pivoting(A, b):
    # Perform QR decomposition with pivoting
    Q, R, P = qr(A, pivoting=True)
    
    # Compute Q^T b
    Qt_b = Q.T @ b.T
    
    # Determine the number of rows in A
    m = A.shape[0]
    
    # Extract the relevant portion of R (first m rows and columns)
    R_tilde = R[:, :m]
    
    # Solve for y in the upper triangular system R_tilde y = Q^T b
    y = np.linalg.solve(R_tilde, Qt_b)
    
    # Construct the solution vector x
    n = A.shape[1]
    x = np.concatenate((y, np.zeros((n - m, 1))))  # Ensure zeros has the same dimension as y

    # Apply the permutation P to y
    x_permuted = np.zeros(n)
    for i, p in enumerate(P):
        x_permuted[p] = x[i]
    
    # Compute the residual
    residual = np.linalg.norm(A @ x_permuted - b)
    
    return x_permuted, residual


# Results for A1
x1, residual1 = solve_qr_pivoting(A1, b)
print("Results for matrix A1: ")
print("--------------------------------")
print("x: \n", x1.T)
print("\nResidual: \n", residual1)

# Results for A2
x2, residual2 = solve_qr_pivoting(A2, b)
print("\nResults for matrix A2: ")
print("--------------------------------")
print("x: \n", x2.T)
print("\nResidual: \n", residual2)
```

### Part (e)

```python
def solve_minimize(A, b, lam):
    At_A = A.T @ A
    At_b = A.T @ b
    M = At_A + lam * np.eye(A.shape[1])
    x = np.linalg.solve(M, At_b)
    residual = np.linalg.norm(A @ x - b)
    return x, residual

# Lambda values to test
lambdas = [0.1, 0.01, 0.001]

# Given matrices A1 and A2 and vector b
A1 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [7, 10, 10, 3, 2, 7]])
A2 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [-2, 1e-5, -8, 7, -3, 6]])
b = np.array([-1, 0, 5])

# Results for A1
print("Results for matrix A1: ")
print("--------------------------------")
for lam in lambdas:
    x1, residual1 = solve_minimize(A1, b, lam)
    print(f"Lambda = {lam}:")
    print("x: \n", x1)
    print("Residual: \n", residual1)
    print()

# Results for A2
print("\nResults for matrix A2: ")
print("--------------------------------")
for lam in lambdas:
    x2, residual2 = solve_minimize(A2, b, lam)
    print(f"Lambda = {lam}:")
    print("x: \n", x2)
    print("Residual: \n", residual2)
    print()
```

```python
import numpy as np
import scipy.linalg as la

# Given matrices A1 and A2 and vector b
A1 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [7, 10, 10, 3, 2, 7]])
A2 = np.array([[5, 4, 1, 8, 4, 8],
               [7, 4, 9, 1, 7, 2],
               [-2, 1e-5, -8, 7, -3, 6]])
b = np.array([-1, 0, 5])

# Perform QR decomposition with pivoting
Q1, R1, P1 = la.qr(A1, pivoting=True)
Q2, R2, P2 = la.qr(A2, pivoting=True)

# Compute Q^T b
Qt_b1 = Q1.T @ b
Qt_b2 = Q2.T @ b

# Get y by solving R y = Q^T b
m = b.shape[0]  # number of rows
y1 = np.linalg.solve(R1[:m, :m], Qt_b1[:m])
y2 = np.linalg.solve(R2[:m, :m], Qt_b2[:m])

# Construct x = P [y; zeros(n-m)]
n1 = A1.shape[1]
n2 = A2.shape[1]
x1 = np.zeros(n1)
x2 = np.zeros(n2)
x1[:m] = y1
x2[:m] = y2

# Apply the permutation P
x1 = x1[np.argsort(P1)]
x2 = x2[np.argsort(P2)]

# Compute the residuals ||Ax - b||
residual1 = np.linalg.norm(A1 @ x1 - b)
residual2 = np.linalg.norm(A2 @ x2 - b)

# Print results
print("Results for matrix A1: ")
print("--------------------------------")
print("x: \n", x1)
print("Residual: \n", residual1)

print("\nResults for matrix A2: ")
print("--------------------------------")
print("x: \n", x2)
print("Residual: \n", residual2)
```
